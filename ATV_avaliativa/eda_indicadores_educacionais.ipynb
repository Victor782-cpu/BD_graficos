Este notebook realiza uma Análise Exploratória de Dados completa, visual e interativa dos indicadores educacionais do Brasil consolidados por macrorregião geográfica. São examinadas métricas cruciais de **fluxo escolar (aprovação, abandono, reprovação)**, **distorção idade-série (TDI)**, **qualificação docente (DSU)** e **disparidades intra-regionais (redes e zonas urbano/rural)**.

```python
# 1. Importação das bibliotecas e carregamento seguro dos dados
from eda_indicadores_educacionais import (
    carregar_e_tratar_dados,
    plot_ranking_regional_aprovacao,
    plot_desigualdade_regional_boxplot,
    plot_mapa_interativo_brasil,
    plot_extremos_top_bottom,
    plot_disparidade_publica_privada,
    plot_correlacao_docentes_aprovacao,
    plot_evolucao_historica_temporal
)

# Carregamento seguro com detecção automática de encoding e separador
df = carregar_e_tratar_dados("br_inep_indicadores_educacionais_regiao.csv")
df.head()
```

## 1. Comparativo Regional: Ranking de Aprovação no Ensino Médio
**Objetivo Pedagógico:** Comparar o desempenho consolidado das cinco macrorregiões brasileiras, ordenadas do melhor para o pior indicador, permitindo uma leitura imediata da eficiência escolar regional.

```python
fig1 = plot_ranking_regional_aprovacao(df, ano_referencia=2024)
fig1.show()
```

## 2. Desigualdade Intra-Regional: Disparidade de Abandono Escolar
**Objetivo Pedagógico:** Evidenciar que as médias agregadas mascaram grandes disparidades. O Box Plot revela a variação interna da evasão escolar entre diferentes redes de ensino (Pública vs Privada) e localização (Urbana vs Rural).

```python
fig2 = plot_desigualdade_regional_boxplot(df, ano_referencia=2024)
fig2.show()
```

## 3. Panorama Geográfico Interativo do Brasil
**Objetivo Pedagógico:** Oferecer uma leitura espacial imediata dos indicadores educacionais sobre o mapa do território brasileiro, integrando múltiplos indicadores em um cartão informativo completo ao passar o cursor.

```python
fig3 = plot_mapa_interativo_brasil(df, ano_referencia=2024)
fig3.show()
```

## 4. Destaques de Extremos: Melhores Cenários vs Gargalos Críticos
